```python
import pandas as pd
```

```python
#Series
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#indexing
print(s[3])
print(s[4])
```

```python
s = pd.Series([23,18,19,24,25], index= ["asha","bob", "nisha","charlie","Ashta"])
print(s["asha"])
print(s["Ashta"])
print(s)

print(s.index)
```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1+s2)

s1[0]= 100
change_s1= s1.drop(0)
print(s1)
print(change_s1)
```

```python
# DataFrame
info = {
    "Name" : ["Adma","Bob","Eve"],
    "GPA" : [9.5,6.5,7.5],
    "Age" : [26,18,25]
}
df = pd.DataFrame(info)
print(df, type(df))
print(df.index)
print(df.columns)
```
