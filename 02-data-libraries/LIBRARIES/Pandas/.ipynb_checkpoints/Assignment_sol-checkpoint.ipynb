```python
import pandas as pd
```

```python
#Que 1
df = pd.read_csv("IRIS.csv")
df2 =df.head(10)
print(df.shape,type(df))
df.columns

df.groupby("sepal_width").max()
df.groupby("sepal_width").min()
df.loc[2]
df[(df["petal_length"]>4.5)]
df[(df["species"]=="Iris-virginica")]
df.groupby("sepal_length").max()
df.groupby("petal_width").min()
df.groupby("species")["sepal_length"].agg(Average_sepal_length = "mean"),
df.groupby("species")["petal_width"].agg(max_petal_width = "mean"),
df.groupby("species")["sepal_width"].agg(std_sepal_width= "std")

df["sepal_length"].agg(["min","max","mean","std"])
```

```python
df["petal_ratio"]= df["petal_length"] / df["petal_width"]
avg=df.groupby("species")["petal_ratio"].mean()
```

```python
df["petal_ratio"] = df["petal_length"] / df["petal_width"]

df["petal_ratio"].agg(["mean", "std", "min", "max"])
df2 = df["sepal_length"].agg(["min","max","mean","std"])
df2
```
