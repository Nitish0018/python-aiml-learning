```python
import numpy as np
import time
```

```python
size = 1_000_000
#python list
py_list = list(range(size))
start = time.time()
sq_time = [x**2 for x in py_list]
end = time.time()
print(f"python list time = {end - start} sec")
# numpy array
np_arr = np.array(py_list)
start = time.time()
sq_array = np_arr ** 2
end = time.time()
print(f"numpy array time = {end - start} sec")
```

```python
#memory
import sys
print(f"pyhton list size = {sys.getsizeof(py_list) * len(py_list)} bytes")
print(f"numpy list size = {np_arr.nbytes} bytes")
```
